# **16.Wordcloud example**

```python
#Ex:1
import matplotlib.pyplot as plt
from wordcloud import WordCloud
text="DataScience having good career path, DataScience is cutting edge technology, DataScience course is existing in nit,Daniel used to teach DataScience, if we learn DataScience then we will get good package"
wc=WordCloud()
wc.generate(text)
plt.figure(figsize=(12,12))
plt.imshow(wc)
plt.axis('off')
plt.show()
```

```python
#Ex:2
import seaborn as sns
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Example text
text = """
data science machine learning python pandas matplotlib seaborn visualization plot
analysis model training evaluation feature engineering regression classification
nlp text processing visualization dashboard jupyter notebook
"""

# Apply seaborn style globally (optional)
sns.set_style("white")
sns.set_context("talk")

# Get a seaborn colormap as a Matplotlib colormap
cmap = sns.color_palette("rocket", as_cmap=True)  # try "viridis", "mako", "rocket", "flare", etc.

# Create the WordCloud
wc = WordCloud(
    width=1200,
    height=600,
    background_color="white",
    collocations=False,   # don't join words with hyphens into bigrams
    prefer_horizontal=0.9,
    max_words=200,
    colormap=cmap
).generate(text)

# Display
plt.figure(figsize=(12,6))
plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.tight_layout()
plt.show()

# Save to file
wc.to_file("wordcloud_seaborn.png")
```

# **17.Sunburst plot example**

```python
#Ex:1
import plotly.express as px
data=px.data.tips()
figure=px.sunburst(data,path=["day","sex"],values="total_bill")
figure.show()
```

```python
#Ex:2 Sales Sunburst Chart
import seaborn as sns
import plotly.express as px
import pandas as pd

data = {
    "Branch": ["A","A","A","B","B","C","C"],
    "Category": ["Food","Food","Electronic","Food","Cosmetic","Electronic","Cosmetic"],
    "Product": ["Bread","Eggs","TV","Juice","Cream","Laptop","Lipstick"],
    "Sales": [1200, 900, 4500, 1300, 800, 5200, 1100]
}

df = pd.DataFrame(data)

fig = px.sunburst(
    df,
    path=["Branch", "Category", "Product"],
    values="Sales",
    color="Category",
)

fig.update_layout(title="Sales Sunburst Chart")
colors = sns.color_palette("flare", 7).as_hex()
fig.update_traces(marker=dict(colors=colors))
fig.show()

```

```python
#Ex:3 Continet-Country-City Sunburst Chart
import plotly.express as px
import pandas as pd

data = {
    "Continent": ["Asia", "Asia", "Asia", "Europe", "Europe"],
    "Country": ["India", "India", "China", "Germany", "France"],
    "City": ["Delhi", "Mumbai", "Beijing", "Berlin", "Paris"],
    "Population": [20, 18, 21, 4, 11]
}

df = pd.DataFrame(data)

fig = px.sunburst(
    df,
    path=["Continent", "Country", "City"],
    values="Population",
    color="Country",
)

fig.update_layout(title="Sunburst: Continent → Country → City")
fig.show()
```

```python
#Ex:1 Barplot using exercise dataset
import matplotlib.pyplot as plt
import seaborn as sns
df=sns.load_dataset("exercise")
#plot a barplot
sns.barplot(x="time",y="pulse",data=df,color="violet")
plt.title("Excerise Time & Pulse for each person")
plt.show()
```
