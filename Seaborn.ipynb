```python
#9.Plot the survival rate of each class
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
df=sns.load_dataset("titanic")
sns.barplot(x='class',y='survived',data=df)
plt.show()

```

```python
#10.plot the the survival rate by gender and class.
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
df=sns.load_dataset('titanic')
result=df.pivot_table('survived',index='sex',columns='class')
print(result)
```

```python
#11.Plot the survival rate by gender, age and class
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
df=sns.load_dataset('titanic')
diff_ages=pd.cut(df['age'],[0,18,80])
result=df.pivot_table('survived',['sex',diff_ages],'class')
print(result)
```

```python
#12.Finding Most of the tickets, using dist plot
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
df=sns.load_dataset('titanic')
sns.distplot(df['fare'],kde=False)
plt.show()


```

```python
#Scatterplot in Seaborn
import seaborn as sns
import matplotlib.pyplot as plt

# Load built-in dataset
data = sns.load_dataset("iris")

# Scatter plot of sepal length vs sepal width
sns.scatterplot(x="sepal_length", y="sepal_width", data=data, hue="species", style="species")

plt.title("Scatter Plot of Sepal Length vs Width")
plt.show()
```

```python
#Histogram plot in seaborn
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset('tips')
sns.histplot(data["total_bill"],bins=20,kde=True,color="orange")
plt.xlabel('Total Bill(₹)')
plt.ylabel('Frequency')
plt.title('Hotelbill')
plt.show()
```

```python
#Taking exercise dataset and load dataset
import seaborn as sns
df=sns.load_dataset("exercise")
print(df)
```

```python
#barplot using exercise dataset
import matplotlib.pyplot as plt
import seaborn as sns
df=sns.load_dataset("exercise")
#plot a barplot
sns.barplot(x="time",y="pulse",data=df,color="violet")
plt.title("Excerise Time & Pulse for each person")
plt.show()
```

```python
#Histogram plot using exercise dataset
import matplotlib.pyplot as plt
import seaborn as sns
data=sns.load_dataset("exercise")
#plot a hist plot
sns.histplot(data["time"],bins=30,kde=True,color="blue")
plt.xlabel("Time")
plt.ylabel("Pulse rate")
plt.title("Excerise Time & Pulse for each person")
plt.show()
```

```python
#scatterplot using exercise dataset
import seaborn as sns
import matplotlib.pyplot as plt
data=sns.load_dataset("exercise")
sns.scatterplot(x="time",y="pulse",data=data,hue="kind",style="kind")
plt.title("Excerise Time & Pulse for each person")
plt.show()

```
